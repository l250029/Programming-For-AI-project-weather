```python
url="https://archive-api.open-meteo.com/v1/archive"
start_date="2016-01-01"
end_date="2025-12-31"
daily_variables=["temperature_2m_mean",
                 "temperature_2m_max",
                "temperature_2m_min",
                 "relative_humidity_2m_mean",
                 "dew_point_2m_mean",
                "precipitation_sum",
                 "rain_sum",
                 "precipitation_hours",
                "wind_speed_10m_max",
                 "wind_speed_10m_mean",
                 "wind_gusts_10m_max",
                "wind_direction_10m_dominant",
                 "weather_code",
                "cloud_cover_mean",
                 "pressure_msl_mean",
                 "sunshine_duration",
                 "daylight_duration",
                 "shortwave_radiation_sum"]
```

```python
all_data=[]
for city in cities:
  parameters={"latitude":city["latitude"],
            "longitude":city["longitude"],
            "end_date":end_date,
            "start_date":start_date,
            "daily":",".join(daily_variables),
            "timezone":"auto"}
  response=requests.get(url,params=parameters)
  print(city["city"], response.status_code)

  if response.status_code != 200:
    print(response.text)
    break
  data=response.json()
  df=pd.DataFrame(data["daily"])
  df["city"]=city["city"]
  df["country"]=city["country"]
  df["longitude"]=city["longitude"]
  df["latitude"]=city["latitude"]
  all_data.append(df)
  print(city["city"], df.shape)
```
